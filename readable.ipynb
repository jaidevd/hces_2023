```python
from tqdm import tqdm
```

```python
len(df)
```

```python
cols_to_match = ['tot_expenditure', 'cons_total_value']

def check(ix, row):
    hhid, item_code = ix
    fdf = dfs[row.level]
    matching = [c for c in cols_to_match if c in fdf][0]
    old = fdf.loc[(hhid, item_code), matching]
    return (old == row.value) or pd.isna(old)

for ix, row in tqdm(df.sample(1000000).iterrows(), total=1000000):
    assert check(ix, row)
```
